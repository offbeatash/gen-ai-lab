# Tools in LangChain

```python
# @tool decorator (simplest way to  define custom tool)

from langchain_core.tools import tool

@tool
def division(a:int, b:int) -> int:
    """Divides two numbers."""
    return a/b
```

```python
print(division.name)
print(division.description)
print(division.args)
```

```python
# async implementation of the tool

@tool
async def multiply(a:int, b:int) -> int:
    """Multiplies two numbers."""
    return a*b
```

```python
from typing import List, Annotated

@tool
def multiply_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[List[int], "List of B values"],
) -> int:
    """Multiplies 'a' by the maximum value in 'b'."""
    return a*max(b)
```

```python
print((multiply_by_max.args))
```

## Structured tools
Provides more configurability than @tool decorator, without requiring much additional code

```python
from langchain_core.tools import StructuredTool

def multiply(a: int, b: int) -> int:
    """Multiply two numbers."""
    return a*b

async def amultiply(a: int, b:int) -> int:
    """Multiply two numbers."""
    return a*b

calculator = StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a": 2, "b": 7}))
print(await calculator.ainvoke({"a": 4, "b": 4}))
```

#1. Wikipedia integration

```python
import wikipedia

wikipedia.set_user_agent(
    "AgenticAI/1.0 (your-email@example.com)"
)

from langchain_community.tools import WikipediaQueryRun
from langchain_community.utilities import WikipediaAPIWrapper

api_wrapper = WikipediaAPIWrapper(
    top_k_results=5,
    doc_content_chars_max=500,
)

tool = WikipediaQueryRun(api_wrapper=api_wrapper)

print(tool.invoke({"query": "LangChain"}))
```
